Bar Plot

```python
import matplotlib.pyplot as plt
x=["part1","part2","part3","part4"]
y=[98,92,86,97]
colors=["red","blue","green","yellow"]

plt.bar(x,y,color=colors,edgecolor="black")
plt.xlabel("parts of harry potter",fontsize=17)
plt.ylabel("popularity",fontsize=17)
plt.title("popularity of different parts of harry potter",fontsize=20)
plt.show
```

Line plot

```python
x=["day1","day2","day3","day4"]
y=[300,400,500,600]
y1=[500,550,600,720]
plt.plot(x,y,color="red",label="week1")
plt.plot(x,y1,color="blue",label="week2")
plt.legend()
```

Scatter Plot

```python
import numpy as np
x=np.random.randint(10,100,50)
y=np.random.randint(10,100,50)
color=np.random.randint(10,100,50)
plt.scatter(x,y,marker="*")
plt.colorbar(cmap="viridis")
```

pie chart

```python
brands=["mi","oneplus","vivo","oppo"]
x=[22,33,30,15]
ex=[0,0,0,0.1]
plt.pie(x,labels=brands,explode=ex)
```

Box Plot

```python
l=[22,46,54,76,32,23,45,64,66,68,97,86,23]
plt.boxplot(l)
```

Histogram

```python
x=[32,48,76,86,97,95,76,34,56,67,95,98,2]
plt.hist(x,bins=20,edgecolor="black",color="pink")
```

violin plot

```python
a=[32,48,76,86,97,95,76,34,56,67,95,98,2]
plt.violinplot(a,showmedians=True)
```

Stem plot

```python
x=[32,48,76,86,97,95,76,34,56,67,95,98,2]
plt.stem(x,linefmt="--",bottom=10,markerfmt="*",orientation="horizontal")
```

stack plot

```python
days=[1,2,3,4,5,6,7]
nop1=[5,10,30,20,35,60,70]
nop2=[10,20,30,40,50,60,70]
nop3=[8,16,24,32,40,48,56]

plt.stackplot(days,nop1,nop2,nop3)
```

step plot

```python
x=["day1","day2","day3","day4"]
y=[30,40,20,45]
plt.step(x,y,where="post",marker="*")
```

working with legends

```python
x=[1,2,3,4,5]
y=[23,45,54,65,77]
y1=[15,34,45,56,66]
plt.figure(figsize=[3,3])
plt.plot(x,y,label="male")
plt.plot(x,y1,label="female")
plt.legend(loc=1,ncols=2,bbox_to_anchor=(0.8,1.2),labelspacing=2)
```

subplot
