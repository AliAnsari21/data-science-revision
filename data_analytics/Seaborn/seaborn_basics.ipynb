line plot

```python
import seaborn as sns
import pandas as pd 
import matplotlib.pyplot as plt

data={"days":[1,2,3,4,5],
      "nop":[50,40,60,33,44]}
df=pd.DataFrame(data)
print(df)
sns.lineplot(data=data,x="days",y="nop")
```

Bar plot

```python
data=sns.load_dataset("tips")
print(data)
sns.barplot(data=data,x="day",y="tip",estimator="median",hue="sex")
```

Histogram

```python
sns.histplot(data=data,x="day",hue="sex",kde="true")
```

scatter plot

```python
sns.scatterplot(data=data,x="total_bill",y="tip",hue="smoker",size="size")
```

Heat Map

```python
gp=data.groupby("day").agg({"tip":"mean"})
sns.heatmap(gp)
```

Count Plot

```python
sns.countplot(data=data,x="sex")
```

violin plot

```python
sns.violinplot(data=data,x="total_bill")
```

Pair Plot

```python
sns.pairplot(data,diag_kind="hist")
```

Strip Plot

```python
sns.stripplot(data=data,x="day",y="total_bill",hue="sex",dodge=True,jitter=0.3)
```

Box Plot

```python
sns.boxplot(data=data,x="tip",y="sex",orient="h")
```

Cat Plot

```python
sns.catplot(data=data,x="day",y="tip",hue="sex",kind="bar")
```

Style And Color In Plots

```python
data1=sns.load_dataset("exercise")
print(data1)
```

```python
sns.set_style(style="ticks")
sns.barplot(data=data1,x="time",y="pulse")
sns.palplot(sns.color_palette("spring"))
```

Multiple Plot

```python
print(data)
a=sns.FacetGrid(data=data,col="day",height=2)
a.map(sns.barplot,data=data,x="day",y="tip")
```

Relational Plot

```python
sns.relplot(data=data,x="tip",y="total_bill",kind="line",col="day",size="smoker")
```

Swarm Plot

```python
sns.swarmplot(data=data,x="day",y="total_bill")
```

KdePlot

```python
sns.kdeplot(data=data,x="total_bill",hue="day",multiple="stack")
```
