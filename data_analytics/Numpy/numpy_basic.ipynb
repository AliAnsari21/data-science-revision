Creating Array, Slicing, Attributes 

```python
#Array
import numpy as np
arr=np.array([10,20,30,40])
print(arr)
```

```python
#slicing
arr=np.array([10,20,30,40])
print(arr[0:3])
```

```python
#attributes
arr1=np.array([[10,20,30],[20,30,40],[30,40,50]])
print(np.shape(arr1))
print(np.size(arr1))
print(np.ndim(arr1))
print(arr1.dtype)
```

Mathematical operation and function on arrays

```python
arr2=np.array([10,20,30,40])
arr3=np.array([30,40,50,60])
print(arr2+arr3)
print(np.add(arr2,arr3))
print(np.subtract(arr2,arr3))
print(np.multiply(arr2,arr3))
print(np.divide(arr2,arr3))
print(np.exp(arr2))
print(np.sqrt(arr3))
print(np.power(arr3,arr2))
```

Combining and Spliting in arrays

```python
#concatenate
arr2=np.array([10,20,30,40])
arr3=np.array([30,40,50,60])
print(np.concatenate([arr2,arr3]))
print(np.hstack([arr2,arr3]))#for horizontal concatenate
print(np.vstack([arr2,arr3]))#for vertical concatenate
```

```python
#slitting
a=np.array([20,30,40,10,20])
print(np.array_split(a,3))
```

Adding and Removing elements in the array

```python
a=np.array([20,30,40,10,50])
print(np.append(a,90))
print(np.insert(a,1,60))
print(np.delete(a,[2]))
```

Search Sort and Filter arrays 

```python
ar=np.array([7,8,9,4])
print(np.sort(ar))
s=np.where(ar==8)
print(s)
fa=[True,False,True,False]
new=ar[fa]
print(new)
```

Aggregate Function in numpy

```python
a=np.array([10,20,30,40])
print(np.sum(a))
print(np.min(a))
print(np.max(a))
print(np.size(a))
print(np.mean(a))
print(np.cumsum(a))
print(np.cumprod(a))
```

```python
#eg of cumprod
a=[100,110,120,130]
b=[2,3,2,4]
price=np.array(a)
quantity=np.array(b)
print(price,quantity)
c=np.cumprod([price,quantity],axis=0)
print(c[1].sum())
```

Statistical function in array

```python
import statistics as stats
baked_food=[200,300,400,500]
a=np.array(baked_food)
print(np.mean(a))
print(np.median(a))
#print(np.mode(a))
print(stats.mode(a))
print(np.std(a))
print(np.var(a))
```

```python
#corelation function
tobacco_consumptions=[30,40,50,60]
deaths=[100,200,300,400]
print(np.corrcoef([tobacco_consumptions,deaths]))
```
