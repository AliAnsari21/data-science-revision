Handling Duplicates value in pandas

```python
print(data1.duplicated()) #to check duplicate(same value)
print(data1["Employee_ID"].duplicated()) #to check duplicate value in specific columns
print(data1["Employee_ID"].duplicated().sum()) #to check no of duplicate value in specific columns
print(data1.drop_duplicates("Employee_ID")) #drop duplicate value through specific columns
```

Working With missing values

```python
import numpy as np
print(data1.isnull().sum) #to find empty values
print(data1.dropna()) #to remove empty or null values
print(data1.replace(np.nan,"hi")) #to relpace null values
data1["Salary"]=data1["Salary"].replace(np.nan,85155.05639558233) # {to replace null value in specific columns}
print(data1["Salary"].mean())
print(data1.ffill())
print(data1.isnull().sum)
```

Columns Transformation Pandas

```python
data1.loc[(data1["Salary"]<=85155.05639558233),"GetBonus"]="no bonus"
data1.loc[(data1["Salary"]>85155.05639558233),"GetBonus"]="bonus"
print(data1.head(10))
```

```python
data1["full_name"]=data1["First_Name"] +" "+ data1["Last_Name"]
data1["full_name"]=data1["First_Name"].str.upper() +" "+ data1["Last_Name"] #to make whole word in capital letter
data1["full_name"]=data1["First_Name"].str.capitalize() +" "+ data1["Last_Name"].str.capitalize()#to make first letter capital
print(data1.head())
```

```python
#making column of 20% bonus
data1["Bonus"]=(data1["Salary"]/100)*20
print(data1.head())
```

```python
#mapping in pandas
data={"months":["January","Febuary","March"]}
a=pd.DataFrame(data)
print(a)
def extract(value):
    return value[0:3]
a["short_months"]=a["months"].map(extract)
print(a)
```

Merge JOIN Concatenate

```python
data2={"emp_id":["E01","E02","E03","E04","E05"],
       "names":["ram","sham","rahul","vishal","lisa"],
       "age":[23,34,43,23,22]}
data3={"emp_id":["E01","E02","E03","E04","E05","e06"],
       "salary":[45000,35000,30000,43000,50000,23000]}

df1=pd.DataFrame(data2)
df2=pd.DataFrame(data3)
print(df1)
print(df2)
```

```python
print(pd.merge(df1,df2))
print(pd.merge(df1,df2,on="emp_id"))
print(pd.merge(df1,df2,on="emp_id",how="right"))
```

Compare DataFrame

```python
dict={"fruits":["mango","apple","banana","orange"],
      "price":[100,80,75,85],
      "quantity":[15,10,10,20]}
df4=pd.DataFrame(dict)
print(df4)
df5=df4.copy()
print(df5)
df5.loc[0,"price"]
df5.loc[1,"price"]
print(df4.compare(df5,align_axis=0))
```

pivoting and melting

```python
print(pd.pivot("fruits","price","quantity"))
```

```python
print(pd.melt(df4,id_vars=["fruits"],value_vars=["quantity"]))
```
