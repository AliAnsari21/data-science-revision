SQL and SQLite

```python
import sqlite3
#connect to sqlite database
connection=sqlite3.connect("example.db")
cursor=connection.cursor()
```

```python
#create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS employee(
 id Integer Primary Key,
 name Text Not Null,
 age Integer,
 department Text
)
""")
```

```python
#commit the change
connection.commit()
cursor.execute('''
Select*from employee
''')
```

```python
#insert the data in sqlite table
cursor.execute('''
Insert Into employee (name,age,department)
values('ALI',22,'Data Science')
''')

cursor.execute('''
Insert Into employee (name,age,department)
values('BOB',25,'Data Engineer')
''')

connection.commit()
```

```python
#query the data from the table
cursor.execute('Select*from employee')
rows=cursor.fetchall()
```

```python
#print the queried data
for row in rows:
    print(row)
```

```python
#update data in table
cursor.execute('''
UPDATE employee
set age = 22
where name = "BOB"
''')

connection.commit()
```

```python
cursor.execute('Select*from employee')
rows=cursor.fetchall()
for row in rows:
    print(row)
```

```python
#delete data from table
cursor.execute('''
Delete from employee
where name = "BOB"
''')
connection.commit()
cursor.execute('Select*from employee')
rows=cursor.fetchall()
for row in rows:
    print(row)
```
